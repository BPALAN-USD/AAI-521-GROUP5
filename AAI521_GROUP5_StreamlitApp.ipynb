```python
import threading
import time
import subprocess

from google.colab import output

# Kill any previous streamlit processes (avoid port conflicts)
!pkill streamlit || echo "no existing streamlit process"

def run_streamlit():
    # launch streamlit
    subprocess.run([
        "streamlit", "run", "streamlit_app.py",
        "--server.port", "8501",
        "--server.headless", "true",
        "--server.enableCORS", "false",
        "--server.enableXsrfProtection", "false"
    ])

# Run Streamlit in the background
thread = threading.Thread(target=run_streamlit, daemon=True)
thread.start()

# Give Streamlit a moment to start
time.sleep(5)

# Get the proxied URL from Colab
url = output.eval_js("google.colab.kernel.proxyPort(8501)")
print("Open this URL in a new tab:")
print(url)
```
